```python
df = pd.merge(data,df_c,on="Country Code", how="left")
```

```python
df.sample(5)
```

```python
country_names = df.Country.value_counts().index
```

```python
country_val = df.Country.value_counts().values
```

```python
# Top 3 countries are used Zomato
plt.pie(country_val[:3],labels=country_names[:3],autopct="%.2f%%")
plt.show()
```

Observations: Zomato have maximum orders or transaction from India After that USA and then UK.

```python
df.columns
```

```python
Ratings = df.groupby(["Aggregate rating","Rating color","Rating text"]).size().reset_index().rename(columns={0:"Rating count"})
```

```python
Ratings
```

Obesrvation: 
1. When Rating is between 4.5 to 4.9 ---> Excellent
2. When Rating is between 4.0 to 4.4 ---> Very Good
3. When Rating is between 3.5 to 3.9 ---> Good
4. When Rating is between 2.5 to 3.4 ---> Avarage
5. When Rating is between 0.1 to 2.4 ---> Poor
6. When Rating is 0 ---> Not rated	

```python
sns.barplot(x = "Aggregate rating", y = "Rating count", data = Ratings)
```

```python
sns.barplot(x = "Aggregate rating", y = "Rating count", data = Ratings,hue="Rating color",palette=["lightblue","Red","Orange","Yellow","green","darkgreen"])
```

Observation:
1. Not Rated count is Very high
2. Maximum number of rating are between 2.5 to 3.4

```python
# Count Plot
sns.countplot(x = "Rating color", data=Ratings,palette=["lightblue","Red","Orange","Yellow","green","darkgreen"])
plt.show()
```
